```python
import hashlib
from datetime import datetime

# REQUIRED: enter your name
student_name = "Enter Your Name Here"

# Time-based component (unique per student, per moment)
timestamp = datetime.now().isoformat(timespec="seconds")

# Convert (name + time) into a reproducible integer seed
seed_string = student_name + "|" + timestamp
seed = int(hashlib.sha256(seed_string.encode()).hexdigest()[:8], 16)

print("Your random seed is:", seed)
print("Timestamp used:", timestamp)

# One RNG for the entire notebook
rng = np.random.default_rng(seed)
```

### Generate the population dataset

This dataset is designed to include:
- continuous measurements (e.g., spending-like quantities)
- discrete counts (e.g., party size)
- a categorical variable (day)

Run the next cell and inspect `df.head()` and `df.describe()`.


```python
n = 800

# Student-specific parameters (makes students' populations meaningfully different)
bill_scale = rng.uniform(6.0, 11.0)
tip_scale = rng.uniform(1.0, 2.6)

# Day distribution (random but reasonable)
day_probs = rng.dirichlet([2, 1, 3, 2])  # sums to 1

df = pd.DataFrame({
    "total_bill": rng.gamma(shape=2.5, scale=bill_scale, size=n),
    "tip": rng.gamma(shape=2.0, scale=tip_scale, size=n),
    "party_size": rng.integers(1, 7, size=n),  # 1..6
    "day": rng.choice(["Thur", "Fri", "Sat", "Sun"], size=n, p=day_probs),
})

df["tip_pct"] = 100 * df["tip"] / df["total_bill"]

print("Population dataset shape:", df.shape)
df.head()
```
