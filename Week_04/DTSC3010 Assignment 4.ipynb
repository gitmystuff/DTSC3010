* Look at the data structure.
* Identify which feature logically represents the requested variable.
* Confirm assumption (e.g., checking the units).

**2. Units and Scale Awareness**

The name `Views_Millions` explicitly communicates the **scale** of the data. If the column was simply named `Total_Views`, we might see the value "70.1" and mistakenly assume 70 people (and 1/10 of a person) watched the show. Including the unit in the header helps us realize that:

* The raw value is $70.1 \times 10^6$.
* The axis labels should reflect these units (e.g., "Views in Millions").
