## Part 1: Scatter Plots - Exploring Relationships

**Scatter plots** help us visualize the relationship between two numerical variables.

Let's look at data on the top 50 highest-grossing actors in Hollywood.

```python
# Load the actors dataset
actors = pd.read_csv('data/actors.csv')

# Display first few rows
actors.head(10)
```

### Understanding the Data

Each row represents one actor with:
- **Total Gross**: Total box office receipts (in millions)
- **Number of Movies**: How many movies they've been in
- **Average per Movie**: Total Gross ÷ Number of Movies
- **#1 Movie**: Their highest-grossing film

### Scatter Plot: Number of Movies vs Total Gross

**Question:** Do actors who appear in more movies earn more at the box office?

```python
# Create a scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(actors['Number of Movies'], actors['Total Gross'], alpha=0.6, s=100)
plt.xlabel('Number of Movies', fontsize=12)
plt.ylabel('Total Gross (millions $)', fontsize=12)
plt.title('Actor Success: Movies vs Box Office Earnings', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**What do we see?**
- The plot slopes **upward** from left to right
- This shows a **positive association**: more movies generally means higher total earnings
- There's variability - some actors with many movies have moderate earnings, others with fewer movies have high earnings

### Scatter Plot: Number of Movies vs Average per Movie

**Question:** Do actors in fewer movies have higher average earnings per film?

```python
# Create another scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(actors['Number of Movies'], actors['Average per Movie'],
            alpha=0.6, s=100, color='coral')
plt.xlabel('Number of Movies', fontsize=12)
plt.ylabel('Average per Movie (millions $)', fontsize=12)
plt.title('Number of Movies vs Average Earnings per Movie', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**What's different here?**
- This plot slopes **downward** - a **negative association**
- More movies generally means *lower* average per movie
- Notice one **outlier**: an actor far to the left with very high average earnings

Let's find that outlier!

```python
# Find actors with fewer than 10 movies
few_movies = actors[actors['Number of Movies'] < 10]
few_movies
```

**Anthony Daniels** - the actor who played C-3PO in Star Wars! His 7 movies were all blockbuster Star Wars films, giving him an incredibly high average.

### Using Seaborn for Enhanced Scatter Plots

Seaborn can create more informative scatter plots with trend lines.

```python
# Scatter plot with regression line
plt.figure(figsize=(10, 6))
sns.regplot(data=actors, x='Number of Movies', y='Total Gross',
            scatter_kws={'alpha':0.6, 's':100},
            line_kws={'color':'red', 'linewidth':2})
plt.title('Number of Movies vs Total Gross (with trend line)',
          fontsize=14, fontweight='bold')
plt.xlabel('Number of Movies', fontsize=12)
plt.ylabel('Total Gross (millions $)', fontsize=12)
plt.tight_layout()
plt.show()
```

## Part 2: Line Graphs - Visualizing Trends Over Time

**Line graphs** are perfect for showing how things change over time.

Let's explore movie industry trends from 1980 to 2015.

```python
# Load movies by year data
movies_by_year = pd.read_csv('data/movies_by_year.csv')
movies_by_year.head(10)
```

### Line Plot: Number of Movies Released Each Year

```python
# Create line plot
plt.figure(figsize=(12, 6))
plt.plot(movies_by_year['Year'], movies_by_year['Number of Movies'],
         marker='o', linewidth=2.5, markersize=4, color='steelblue')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Number of Movies Released', fontsize=12)
plt.title('Movie Production Over Time (1980-2015)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Observations:**
- Sharp rise in the early 1980s
- Gradual upward trend through 2015

### Focusing on Recent Years (2000-2015)

## Part 4: Histograms - Understanding Distributions

**Histograms** show us how numerical data is distributed across different ranges (bins).

**Key difference from bar charts:**
- Bar charts: categorical data (like ice cream flavors)
- Histograms: numerical data grouped into ranges

### Example: Movie Earnings Distribution

Let's create sample data for top movies' adjusted gross earnings.

```python
# Create sample movie earnings data (in millions, adjusted for inflation)
np.random.seed(42)
adjusted_gross = np.concatenate([
    np.random.uniform(300, 400, 68),  # Most movies in 300-400 range
    np.random.uniform(400, 500, 60),  # Good number in 400-500
    np.random.uniform(500, 600, 32),  # Fewer in 500-600
    np.random.uniform(600, 700, 15),  # Even fewer higher
    np.random.uniform(700, 900, 14),
    np.random.uniform(900, 1200, 6),
    np.random.uniform(1200, 1800, 5)  # Just a few blockbusters
])

movies_data = pd.DataFrame({
    'Adjusted Gross': adjusted_gross
})

print(f"We have data on {len(movies_data)} movies")
print(f"Range: ${movies_data['Adjusted Gross'].min():.1f}M to ${movies_data['Adjusted Gross'].max():.1f}M")
```

### Creating a Histogram with Matplotlib

```python
# Create histogram with specified bins
plt.figure(figsize=(12, 6))
bins = range(300, 2001, 100)  # Bins: 300-400, 400-500, etc.
plt.hist(movies_data['Adjusted Gross'], bins=bins,
         color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Adjusted Gross (Million Dollars)', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)
plt.title('Distribution of Movie Earnings', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**What we see:**

- Most movies earned between $300-500 million

- Fewer movies in higher earnings ranges

- The distribution is **skewed to the right** (long tail on the right side)

- A small number of blockbusters earned over $1 billion

### Using Seaborn for More Informative Histograms

```python
# Seaborn histogram with density curve
plt.figure(figsize=(12, 6))
sns.histplot(data=movies_data, x='Adjusted Gross', bins=bins,
             kde=True, color='purple', alpha=0.6)
plt.xlabel('Adjusted Gross (Million Dollars)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Distribution of Movie Earnings (with density curve)',
          fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

The smooth curve (KDE - Kernel Density Estimate) shows the overall pattern of the distribution.

### Comparing Histogram Bin Sizes

Different bin widths reveal different levels of detail.

```python
# Create three histograms with different bin sizes
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Coarse bins (few large bins)
axes[0].hist(movies_data['Adjusted Gross'], bins=5,
             color='lightcoral', edgecolor='black', alpha=0.7)
axes[0].set_title('Coarse Bins (5 bins)', fontweight='bold')
axes[0].set_xlabel('Adjusted Gross (Million $)')
axes[0].set_ylabel('Count')
axes[0].grid(axis='y', alpha=0.3)

# Medium bins
axes[1].hist(movies_data['Adjusted Gross'], bins=range(300, 2001, 100),
             color='steelblue', edgecolor='black', alpha=0.7)
axes[1].set_title('Medium Bins (17 bins)', fontweight='bold')
axes[1].set_xlabel('Adjusted Gross (Million $)')
axes[1].grid(axis='y', alpha=0.3)

# Fine bins (many small bins)
axes[2].hist(movies_data['Adjusted Gross'], bins=30,
             color='lightgreen', edgecolor='black', alpha=0.7)
axes[2].set_title('Fine Bins (30 bins)', fontweight='bold')
axes[2].set_xlabel('Adjusted Gross (Million $)')
axes[2].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

**Notice:**
- Too few bins: lose detail
- Too many bins: too noisy
- Medium bins: just right for seeing the overall pattern!

Sometimes we want to compare multiple datasets on the same plot.

## Summary: Choosing the Right Visualization

| Visualization Type | Best Used For | Example |
|-------------------|---------------|----------|
| **Scatter Plot** | Relationships between two numerical variables | Movies vs Earnings |
| **Line Graph** | Trends over time | Movies released per year |
| **Bar Chart** | Comparing categories | Studios' movie counts |
| **Histogram** | Distribution of numerical data | Movie earnings ranges |
| **Overlaid Graphs** | Comparing multiple groups | USA vs California demographics |

## Key Takeaways

1. **Scatter plots** reveal associations:
   - Positive: both variables increase together
   - Negative: one increases as the other decreases

2. **Line graphs** show trends over time - perfect for time series data

3. **Bar charts** compare categories - bars can be reordered for clarity

4. **Histograms** show distributions:
   - Bars represent ranges (bins), not individual values
   - Width and height both matter
   - Can reveal skewness and outliers

5. **Overlaying** allows direct comparison - but don't overcrowd!

6. **Always label** your axes and include titles

## 1. Defining Functions

Functions allow us to give a name to a computational process that we can use repeatedly.

Let's start with a simple example:

```python
# Our first function definition
def double(x):
    """Double x"""
    return 2 * x

# Test it
print(double(17))
print(double(-0.6/4))
```

Functions can also work with arrays:

```python
# Function works with arrays too
numbers = np.array([3, 4, 5])
print(double(numbers))
```

### Functions with Multiple Arguments

Let's create a more useful function:

```python
def percent(x, total):
    """Convert x to a percentage of total.

    More precisely, this function divides x by total,
    multiplies the result by 100, and rounds the result
    to two decimal places.

    Examples:
    >>> percent(4, 16)
    25.0
    >>> percent(1, 6)
    16.67
    """
    return round((x/total)*100, 2)

# Test it
print(percent(33, 200))
```

```python
# Function that works with arrays
def percents(counts):
    """Convert the values in array to percents out of the total."""
    total = counts.sum()
    return np.round((counts/total)*100, 2)

# Test it
some_array = np.array([7, 10, 4])
print(percents(some_array))
```

### Functions with Default Arguments

```python
def percents_flexible(counts, decimal_places=2):
    """Convert values to percents with flexible decimal places."""
    total = counts.sum()
    return np.round((counts/total)*100, decimal_places)

parts = np.array([2, 1, 4])
print("Rounded to 1 decimal place:", percents_flexible(parts, 1))
print("Rounded to 2 decimal places:", percents_flexible(parts, 2))
print("Rounded to default:", percents_flexible(parts))
```

## 2. Applying Functions to DataFrame Columns

Let's create a DataFrame and learn how to apply functions to its columns.

```python
# Create a simple DataFrame
ages = pd.DataFrame({
    'Person': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Age': [17, 117, 52, 100, 6, 101]
})

ages
```

Now let's define a function to cut off ages at 100:

```python
def cut_off_at_100(x):
    """The smaller of x and 100"""
    return min(x, 100)

# Test the function
print(cut_off_at_100(17))
print(cut_off_at_100(117))
print(cut_off_at_100(100))
```

### Using `.apply()` to Transform Columns

```python
# Apply the function to the Age column
cut_off_ages = ages['Age'].apply(cut_off_at_100)
print(cut_off_ages)
```

```python
# Add as a new column
ages['Cut Off Age'] = ages['Age'].apply(cut_off_at_100)
ages
```

### Example: Height Prediction

Let's work with parent and child height data:

```python
# Create sample height data
# (In practice, you'd load this from the family_heights.csv file)
np.random.seed(42)
n = 200
father_heights = np.random.normal(69, 3, n)
mother_heights = np.random.normal(64, 2.5, n)
child_heights = (father_heights + mother_heights) / 2 + np.random.normal(0, 3, n)

parent_avg = (father_heights + mother_heights) / 2

heights = pd.DataFrame({
    'Parent Average': parent_avg,
    'Child': child_heights
})

heights.head()
```

```python
# Scatter plot of the data
plt.figure(figsize=(10, 6))
plt.scatter(heights['Parent Average'], heights['Child'], alpha=0.5)
plt.xlabel('Parent Average Height (inches)')
plt.ylabel('Child Height (inches)')
plt.title('Parent Average vs Child Height')
plt.grid(True, alpha=0.3)
plt.show()
```

Now let's create a prediction function:

```python
def predict_child(p_avg, df=heights, window=0.5):
    """Predict the height of a child whose parents have a parent average
    height of p_avg.

    The prediction is the average height of the children whose parent
    average height is in the range p_avg plus or minus window.
    """
    # Filter to rows where Parent Average is close to p_avg
    close_points = df[(df['Parent Average'] >= p_avg - window) &
                      (df['Parent Average'] <= p_avg + window)]
    return close_points['Child'].mean()

# Test it
print(f"Predicted height for parent average of 68: {predict_child(68):.2f}")
print(f"Predicted height for parent average of 66: {predict_child(66):.2f}")
```

```python
# Apply prediction to all parent averages
heights['Prediction'] = heights['Parent Average'].apply(predict_child)
heights.head()
```

```python
# Plot the data with predictions
plt.figure(figsize=(10, 6))
plt.scatter(heights['Parent Average'], heights['Child'], alpha=0.5, label='Actual')
plt.scatter(heights['Parent Average'], heights['Prediction'],
           color='gold', s=30, alpha=0.7, label='Prediction')
plt.xlabel('Parent Average Height (inches)')
plt.ylabel('Height (inches)')
plt.title('Height Predictions Based on Parent Average')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Example: Education and Income

Let's create sample data about education and income:

```python
# Create sample education/income data
np.random.seed(42)

education_levels = [
    'No high school diploma',
    'High school or equivalent',
    'College, less than 4-yr degree',
    "Bachelor's degree or higher"
]

income_levels = [
    'A: 0 to 4,999',
    'B: 5,000 to 9,999',
    'C: 10,000 to 14,999',
    'D: 15,000 to 24,999',
    'E: 25,000 to 34,999',
    'F: 35,000 to 49,999',
    'G: 50,000 to 74,999',
    'H: 75,000 and over'
]

# Generate sample data
n_records = 1000
educ_inc = pd.DataFrame({
    'Educational Attainment': np.random.choice(education_levels, n_records,
                                               p=[0.16, 0.23, 0.29, 0.32]),
    'Personal Income': np.random.choice(income_levels, n_records)
})

educ_inc.head()
```

```python
# Count by education level
education_counts = educ_inc['Educational Attainment'].value_counts()
education_counts
```

```python
# Convert to percentages
education_pct = (education_counts / education_counts.sum() * 100).round(2)
print("Distribution of Educational Attainment (%)")
print(education_pct)
```

## 5. Joining DataFrames

Often data about the same items is stored in different tables. Let's see how to combine them:

```python
# Ice cream cones table
cones = pd.DataFrame({
    'Flavor': ['strawberry', 'vanilla', 'chocolate', 'strawberry', 'chocolate'],
    'Price': [3.55, 4.75, 6.55, 5.25, 5.75]
})

# Ratings table
ratings = pd.DataFrame({
    'Kind': ['strawberry', 'chocolate', 'vanilla'],
    'Stars': [2.5, 3.5, 4]
})

print("Cones:")
print(cones)
print("\nRatings:")
print(ratings)
```

### Using `.merge()` to Join

```python
# Merge (join) the tables
rated = cones.merge(ratings, left_on='Flavor', right_on='Kind')
rated = rated.drop('Kind', axis=1)  # Remove duplicate column
rated
```

```python
# Calculate price per star (value metric)
rated['$Star'] = (rated['Price'] / rated['Stars']).round(2)
rated = rated.sort_values('$Star')
rated
```

### Example with Reviews

```python
# Reviews data
reviews = pd.DataFrame({
    'Flavor': ['vanilla', 'chocolate', 'vanilla', 'chocolate'],
    'Stars': [5, 3, 5, 4]
})

# Calculate average review per flavor
average_review = reviews.groupby('Flavor')['Stars'].mean().reset_index()
average_review.columns = ['Flavor', 'Stars_average']
print(average_review)
```

```python
# Join with cones
cones_with_reviews = cones.merge(average_review, on='Flavor')
cones_with_reviews
```

Notice that strawberry disappeared - there were no strawberry reviews!

## 6. Real-World Example: Bay Area Bike Share

Let's create a sample bike sharing dataset similar to the Bay Area Bike Share:

```python
# Create sample bike trip data
np.random.seed(42)
n_trips = 1000

stations = [
    'San Francisco Caltrain (Townsend at 4th)',
    'San Francisco Caltrain 2 (330 Townsend)',
    'Harry Bridges Plaza (Ferry Building)',
    'Embarcadero at Sansome',
    'Market at Sansome',
    'Steuart at Market',
    '2nd at Townsend',
    'Townsend at 7th'
]

# Generate trip durations (in seconds)
durations = np.random.gamma(5, 120, n_trips)  # Most around 10 min, some longer
durations = np.clip(durations, 60, 1800)  # Clip to 1 min - 30 min

trips = pd.DataFrame({
    'Duration': durations.astype(int),
    'Start Station': np.random.choice(stations, n_trips),
    'End Station': np.random.choice(stations, n_trips),
    'Subscriber Type': np.random.choice(['Subscriber', 'Customer'], n_trips, p=[0.8, 0.2])
})

trips.head(10)
```

### Exploring Trip Durations

```python
# Filter to trips under 30 minutes (1800 seconds)
commute = trips[trips['Duration'] < 1800].copy()
print(f"Total trips: {len(trips)}")
print(f"Trips under 30 min: {len(commute)}")
```

```python
# Plot histogram of durations
plt.figure(figsize=(10, 6))
plt.hist(commute['Duration'], bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Duration (seconds)')
plt.ylabel('Frequency')
plt.title('Distribution of Bike Trip Durations')
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

### Most Popular Start Stations

```python
# Count trips by start station
starts = commute.groupby('Start Station').size().reset_index(name='count')
starts = starts.sort_values('count', ascending=False)
starts
```

```python
# Visualize top stations
plt.figure(figsize=(12, 6))
plt.barh(starts['Start Station'], starts['count'], color='coral')
plt.xlabel('Number of Trips')
plt.ylabel('Station')
plt.title('Most Popular Start Stations')
plt.tight_layout()
plt.show()
```

### Cross-Classification: Start and End Stations

```python
# Group by both start and end stations
routes = commute.groupby(['Start Station', 'End Station']).size().reset_index(name='count')
routes = routes.sort_values('count', ascending=False)
routes.head(10)
```

```python
# Create a pivot table (crosstab) for routes
route_pivot = pd.crosstab(commute['Start Station'], commute['End Station'])
route_pivot
```

### Visualizing with a Heatmap

```python
# Create heatmap of routes
plt.figure(figsize=(12, 10))
sns.heatmap(route_pivot, annot=True, fmt='d', cmap='YlOrRd', cbar_kws={'label': 'Number of Trips'})
plt.title('Bike Trip Routes: Start to End Stations')
plt.xlabel('End Station')
plt.ylabel('Start Station')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
