# Section 1: Permutations and Combinations

## Learning Objectives
- Understand when order matters (permutations) vs. when it doesn't (combinations)
- Calculate probabilities using counting methods
- Apply these concepts to real-world scenarios

## Background

### Permutations (Order Matters)
When order matters (like a lock combination or race rankings), we use **permutations**.

Formula: $nPr = \frac{n!}{(n-r)!}$

Example: How many ways can we arrange 3 books from a shelf of 5?
- $5P3 = \frac{5!}{(5-3)!} = \frac{5!}{2!} = 5 \times 4 \times 3 = 60$

### Combinations (Order Doesn't Matter)
When order doesn't matter (like pizza toppings or team selection), we use **combinations**.

Formula: $nCr = \binom{n}{r} = \frac{n!}{r!(n-r)!}$

Example: How many ways can we choose 3 toppings from 5?
- $5C3 = \frac{5!}{3!2!} = 10$

## Example 1.1: Understanding the Difference

```python
# Example: Selecting 2 students from 4 students: Alice, Bob, Carol, Dave
students = ['Alice', 'Bob', 'Carol', 'Dave']

# PERMUTATIONS: Order matters (President and Vice President)
print("Permutations (President, Vice President):")
perm_list = list(permutations(students, 2))
for p in perm_list:
    print(f"  President: {p[0]}, Vice President: {p[1]}")
print(f"Total permutations: {len(perm_list)}")

print("\n" + "="*50 + "\n")

# COMBINATIONS: Order doesn't matter (Team of 2)
print("Combinations (Team of 2):")
comb_list = list(combinations(students, 2))
for c in comb_list:
    print(f"  Team: {c[0]} and {c[1]}")
print(f"Total combinations: {len(comb_list)}")
```

# Section 2: Conditional Probability and Dependence

## Learning Objectives
- Calculate conditional probabilities
- Distinguish between independent and dependent events
- Apply probability rules correctly

## Background

### Conditional Probability

$P(A|B) = \frac{P(A \cap B)}{P(B)}$

Read as: "The probability of A given B"

### Independence
Events A and B are **independent** if:
- $P(A \cap B) = P(A) \times P(B)$
- Knowing B doesn't change the probability of A

### Dependence
Events are **dependent** if knowing one affects the probability of the other:
- $P(A \cap B) = P(A) \times P(B|A)$

## Background

### Bayes' Theorem

$P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$

Or more specifically for disease testing:

$P(\text{Disease}|\text{Positive}) = \frac{P(\text{Positive}|\text{Disease}) \times P(\text{Disease})}{P(\text{Positive})}$

Where:
- **Sensitivity**: P(Positive Test | Have Disease) - True Positive Rate
- **Specificity**: P(Negative Test | No Disease) - True Negative Rate
- **Prevalence**: P(Disease) - How common is the disease?

### Key Insight
A positive test result doesn't always mean you have the disease! It depends on:
1. How accurate the test is
2. How rare the disease is
