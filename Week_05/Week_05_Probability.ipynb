## Example: Betting on Dice

Suppose you bet on a die roll with these rules:

- Roll 1 or 2: lose $1

- Roll 3 or 4: no change ($0)

- Roll 5 or 6: win $1

```python
def bet_on_die(roll):
    """
    Returns the net gain from betting on a die roll.

    Parameters:
    roll (int): The result of a die roll (1-6)

    Returns:
    int: Net gain in dollars (-1, 0, or 1)
    """
    if roll <= 2:
        return -1
    elif roll <= 4:
        return 0
    else:
        return 1

# Test all possible outcomes
for i in range(1, 7):
    result = bet_on_die(i)
    print(f"Roll {i}: ${result}")
```

## Probability Analysis: Die Roll Bet

## Is this a fair game?
**Yes**, this is a mathematically **fair game**.

In probability theory, a game is considered "fair" if the **Expected Value ($E$)** is equal to **0**. This indicates that over a long series of trials, a player should expect to neither net a profit nor a loss.

### Expected Value Calculation
The expected value is the sum of all possible outcomes multiplied by their respective probabilities:

$$E = \sum (x \cdot P(x))$$

$$E = (-1 \cdot \frac{1}{3}) + (0 \cdot \frac{1}{3}) + (1 \cdot \frac{1}{3})$$

$$E = -\frac{1}{3} + 0 + \frac{1}{3} = 0$$

## Conclusion
The game is fair because the probability of winning is identical to the probability of losing (approximately $33.3\%$ each), and the dollar amounts for both outcomes are equal.

### What would make it fair?
Since the game is **already fair**, no adjustments are required. However, if you were to change the "win" conditions, you would have to adjust the payouts to maintain an expected value of zero.

**Example of an adjusted fair game:**
* **New Rules:** Roll 1–4: Lose $1 | Roll 5–6: Win $2.
* **The Math:** $$E = (-1 \cdot \frac{4}{6}) + (2 \cdot \frac{2}{6})$$
$$E = -\frac{2}{3} + \frac{2}{3} = 0$$

## Expected Value - The Foundation of Decision Making

### What is Expected Value?

**Expected Value** (E[X] or μ) is the long-run average outcome if you repeat an experiment infinitely many times. It's the center of gravity of a probability distribution.

In the simplest terms: the expected value is the sum of each possible outcome multiplied by the probability of 그 outcome occurring.

**Formula**: E[X] = Σ(x × P(x)) for all possible values x

### Example: A Simple Game

You roll a fair six-sided die:
- If you roll 6, you win $10

- If you roll 1-5, you win $0

Should you play if it costs $2 to play?

Let's calculate!

```python
# Calculate expected value of the die game
outcomes = [0, 0, 0, 0, 0, 10]  # outcomes for rolling 1,2,3,4,5,6
probabilities = [1/6, 1/6, 1/6, 1/6, 1/6, 1/6]

# Expected value formula: sum of (outcome * probability)
expected_value = sum([outcome * prob for outcome, prob in zip(outcomes, probabilities)])

print(f"Expected winnings per game: ${expected_value:.2f}")
print(f"Cost to play: $2.00")
print(f"Expected profit: ${expected_value - 2:.2f}")
print(f"\nShould you play? {'Yes!' if expected_value > 2 else 'No!'}")
```

# Part 5: The Simulation Process

**Simulation** is using a computer to mimic a physical random process. Here's our systematic approach:

## The 4-Step Process:

1. **What to Simulate**: Decide what quantity you want to simulate
2. **Simulating One Value**: Write code to simulate one instance
3. **Number of Repetitions**: Decide how many times to repeat
4. **Simulating Multiple Values**: Use a loop to collect all results

### Example: Number of Heads in 100 Coin Tosses

**Question**: In 100 coin tosses, we expect about 50 heads. But how variable is this? How often do we get exactly 50?

```python
# Step 1: What to simulate?
# We want to simulate the NUMBER OF HEADS in 100 tosses

# Step 2: Simulating one value
def simulate_100_tosses():
    """
    Simulate 100 coin tosses and count the heads.

    Returns:
    int: Number of heads in 100 tosses
    """
    coin = np.array(['Heads', 'Tails'])
    tosses = np.random.choice(coin, size=100)
    num_heads = np.sum(tosses == 'Heads')
    return num_heads

# Test it once
print(f"Number of heads in 100 tosses: {simulate_100_tosses()}")
```

```python
# Step 3: Number of repetitions
num_repetitions = 10000

# Step 4: Simulate multiple values
results = []

for i in range(num_repetitions):
    num_heads = simulate_100_tosses()
    results.append(num_heads)

results = np.array(results)
print(f"Completed {num_repetitions} simulations!")
print(f"First 10 results: {results[:10]}")
```

### Visualizing the Results

```python
# Create a histogram
plt.figure(figsize=(12, 6))
plt.hist(results, bins=range(30, 71), edgecolor='black', alpha=0.7)
plt.xlabel('Number of Heads in 100 Tosses')
plt.ylabel('Frequency')
plt.title(f'Distribution of Heads in 100 Coin Tosses ({num_repetitions} simulations)')
plt.axvline(50, color='red', linestyle='--', linewidth=2, label='Expected Value (50)')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()

# Summary statistics
print(f"Mean: {np.mean(results):.2f}")
print(f"Standard Deviation: {np.std(results):.2f}")
print(f"Exactly 50 heads: {np.sum(results == 50)} times ({100*np.sum(results == 50)/num_repetitions:.1f}%)")
```

## The Binomial Distribution

### 1. Verification of the Distribution
This is a textbook example of a **Binomial Distribution**.

Since you are dealing with a fixed number of independent trials (100 coin tosses) where each trial has only two possible outcomes (Heads or Tails) and a constant probability of success ($p = 0.5$), the results naturally follow a binomial model.

#### Why it fits the "BINS" Criteria:
* **Binary:** There are only two outcomes (Heads or Tails).
* **Independent:** One coin toss does not affect the outcome of the next.
* **Number:** There is a fixed number of trials ($n=100$).
* **Success:** The probability of "success" (Heads) remains constant ($p=0.5$).

As $n$ increases, the Binomial distribution begins to look like a bell curve. This is the **Normal Approximation**, which occurs because the number of trials is large enough for the **Central Limit Theorem** to take effect.

### 2. Analysis

**What do you notice about the shape of this distribution?**
The distribution is **symmetric, unimodal, and bell-shaped**. It is centered at 50, which is the expected value ($n \times p$). The symmetry indicates that being 5 units above the mean is just as likely as being 5 units below it.

**How often do we get exactly 50 heads?**
According to your simulation, it happened **797 times (8.0%)**. While 50 is the most likely single outcome, the probability of hitting it *exactly* is relatively low because the total probability (100%) is spread across many likely outcomes (like 48, 49, 51, and 52).

**What range contains most of the results?**
Most results fall within one standard deviation ($\sigma \approx 5$) of the mean.
* **68% of results:** Roughly between **45 and 55**.
* **95% of results:** Roughly between **40 and 60** (two standard deviations).
Anything outside the 40–60 range would be considered quite rare in this simulation.

### Example: Coin Flipping

Let's flip a fair coin 10 times and count the heads.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import binom, norm, gaussian_kde
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Simulate flipping a fair coin 10 times
n_flips = 10
p_heads = 0.5

# Simulate many sets of 10 flips
n_simulations = 10000
results = []

for _ in range(n_simulations):
    flips = np.random.binomial(n=n_flips, p=p_heads)
    results.append(flips)

# Plot the distribution
plt.figure(figsize=(10, 6))
plt.hist(results, bins=range(0, n_flips+2), density=True, alpha=0.7, edgecolor='black', label='Simulated')

# Overlay theoretical binomial distribution
x = range(0, n_flips+1)
theoretical_probs = [binom.pmf(k, n_flips, p_heads) for k in x]
plt.plot(x, theoretical_probs, 'ro-', linewidth=2, markersize=8, label='Theoretical')

plt.xlabel('Number of Heads')
plt.ylabel('Probability')
plt.title(f'Binomial Distribution: n={n_flips}, p={p_heads}')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print(f"Average number of heads in {n_simulations} simulations: {np.mean(results):.2f}")
print(f"Expected value (n × p): {n_flips * p_heads}")
```

# Part 6: Rolling Two Dice (Monopoly Moves)

In Monopoly, you roll two dice and move the sum. What should you expect?

```python
# Step 1 & 2: Simulate one Monopoly move
def one_monopoly_move():
    """
    Simulate rolling two dice and return their sum.

    Returns:
    int: Sum of two dice (2-12)
    """
    die = np.arange(1, 7)
    roll1 = np.random.choice(die)
    roll2 = np.random.choice(die)
    return roll1 + roll2

# Step 3 & 4: Simulate many moves
num_simulations = 10000
monopoly_moves = []

for i in range(num_simulations):
    total = one_monopoly_move()
    monopoly_moves.append(total)

monopoly_moves = np.array(monopoly_moves)
```

```python
# Visualize
plt.figure(figsize=(12, 6))
plt.hist(monopoly_moves, bins=range(2, 14), edgecolor='black', alpha=0.7, density=True)
plt.xlabel('Sum of Two Dice')
plt.ylabel('Probability')
plt.title(f'Distribution of Monopoly Moves ({num_simulations} simulations)')
plt.xticks(range(2, 13))
plt.grid(axis='y', alpha=0.3)
plt.show()

# Most common roll
values, counts = np.unique(monopoly_moves, return_counts=True)
most_common = values[np.argmax(counts)]
print(f"Most common sum: {most_common}")
print(f"Average sum: {np.mean(monopoly_moves):.2f}")
```

# Part 7: A Brief History of Probability

### The Birth of Probability Theory

Probability didn't emerge from pure mathematics—it came from **gambling**!

#### Gerolamo Cardano (1501-1576)
- Italian mathematician and physician
- **Used gambling winnings to pay for medical school**
- Wrote *The Book on Games of Chance* (first systematic treatment of probability)
- His key insight: *"Suppose a random process has many equally likely outcomes, some favorable (winning), some unfavorable (losing). Then the probability of obtaining a favorable outcome is equal to the proportion of outcomes that are favorable."*

#### Galileo Galilei (1564-1642)
- Wrote *Thoughts on the Game of Dice*
- Solved the puzzle: Why does 10 appear more often than 9 when rolling three dice?
- Insight: Count the **arrangements**, not just the sums!

#### Blaise Pascal (1623-1662)
- Corresponded with Pierre de Fermat about gambling problems
- Created **Pascal's Triangle** (useful for counting combinations)
- Introduced the concept of **expected value**
- Had a mystical vision and denounced mathematics... but his work survives!

**Key Takeaway**: The mathematics of randomness was developed by people trying to win at games and understand uncertainty.

```python
import matplotlib.pyplot as plt
from scipy.special import comb

def plot_pascal_triangle(rows):
    plt.figure(figsize=(12, 8))

    for n in range(rows):
        for k in range(n + 1):
            # Calculate the value using nCr
            value = int(comb(n, k))

            # Coordinate logic:
            # y-axis is negative n to flip the triangle downward
            # x-axis is k minus half the row length to center it
            x = k - n / 2
            y = -n

            # Add text to the plot
            plt.text(x, y, str(value),
                     ha='center', va='center',
                     fontsize=12, fontweight='bold',
                     bbox=dict(facecolor='skyblue', alpha=0.3, edgecolor='navy', boxstyle='round,pad=0.5'))

    # Aesthetic adjustments
    plt.title(f"Pascal's Triangle (First {rows} Rows)", fontsize=16)
    plt.axis('off') # Hide the grid and axes
    plt.ylim(-rows, 1)
    plt.xlim(-rows/2 - 1, rows/2 + 1)
    plt.show()

# Change this number to grow or shrink the triangle
plot_pascal_triangle(rows=7)
```

# Case Study: The Drunkard's Walk and the 1996 World Series

The binomial probability of the 1996 World Series. In this scenario, the Atlanta Braves were up 2-0. To win the best-of-seven series, the New York Yankees needed to win **4 out of the remaining 5 games**.

### Theoretical Framework: Pascal's Triangle
Using the "BINS" criteria for a binomial distribution, we assume each game is an independent trial with a $p=0.5$ chance of winning. The number of ways the Yankees can win $k$ games out of $n=5$ remaining is found in **Row 5 of Pascal's Triangle**:

**1, 5, 10, 10, 5, 1**



* **1 way** to win 0 games
* **5 ways** to win 1 game
* **10 ways** to win 2 games
* **10 ways** to win 3 games
* **5 ways** to win 4 games (**Yankees win series**)
* **1 way** to win 5 games (**Yankees win series**)

Total possible outcomes: $2^5 = 32$.
Total winning outcomes for Yankees: $5 + 1 = 6$.
**Probability:** $6/32 = 18.75\%$.

| Yankees Wins | Count | Possible Sequences (Permutations) | Series Winner |
| :--- | :--- | :--- | :--- |
| **0 Wins** | 1 | BBBBB | Braves |
| **1 Win** | 5 | YBBBB, BYBBB, BBYBB, BBBYB, BBBBY | Braves |
| **2 Wins** | 10 | YYBBB, YBYBB, YBBYB, YBBBY, BYYBB, BYBYB, BYBBY, BBYYB, BBYBY, BBBYY | Braves |
| **3 Wins** | 10 | YYYBB, YYBYB, YYBBY, YBYYB, YBYBY, YBBYY, BYYYB, BYYBY, BYBYY, BBYYY | Braves |
| **4 Wins** | 5 | YYYYB, YYYBY, YYBYY, YBYYY, BYYYY | **Yankees** |
| **5 Wins** | 1 | YYYYY | **Yankees** |

```python
import numpy as np
import pandas as pd
from scipy.special import comb

# Parameters
n_remaining = 5
p_win = 0.5
total_outcomes = 2**n_remaining

# Calculate ways and probabilities
ways = [comb(n_remaining, k) for k in range(n_remaining + 1)]
probs = [w / total_outcomes for w in ways]

# Create DataFrame
df_series = pd.DataFrame({
    "Yankee Wins": list(range(n_remaining + 1)),
    "Ways (Pascal Row 5)": ways,
    "Probability": probs,
    "Outcome": ["Braves Win", "Braves Win", "Braves Win", "Braves Win", "Yankees Win", "Yankees Win"]
})

print(df_series.to_string(index=False))

yanks_total_prob = df_series[df_series["Outcome"] == "Yankees Win"]["Probability"].sum()
print(f"\nTotal Probability of Yankees winning the series: {yanks_total_prob:.2%}")
```

### Defining Probability via Sample Space

In probability theory, we use the **Sample Space** as the foundation for all calculations.

#### 1. The Definitions
* **Sample Space ($S$):** The set of all possible outcomes of a random experiment. In our 1996 World Series example, the sample space $S$ consists of the **32** possible sequences of wins and losses ($2^5$).
* **Event ($E$):** A subset of the sample space. This represents the specific outcome(s) we are interested in. For example, "The Yankees win the series" is an event $E$ containing the **6** sequences where they win 4 or 5 games.
* **Outcome ($s$):** A single element within the sample space (e.g., `YYYYB`).

#### 2. The Probability Formula (Classical Definition)
If all outcomes in a finite sample space are equally likely, the probability of an event $E$ occurring is the ratio of the number of favorable outcomes to the total number of outcomes in the sample space:

$$P(E) = \frac{|E|}{|S|} = \frac{\text{Number of outcomes in } E}{\text{Total number of outcomes in } S}$$

#### 3. Application: The 1996 World Series
Using the list we generated:
* $|S| = 32$ (Total possible paths)
* $|E| = 6$ (Paths where Yankees win 4 or 5 games)

$$P(\text{Yankees Win}) = \frac{6}{32} = 0.1875 \text{ (or 18.75%)}$$

### The Three Axioms of Probability
For any sample space, these three rules must hold true:
1.  **Non-negativity:** The probability of any event is at least 0 ($P(E) \geq 0$).

**The Surprising Answer**

**You should SWITCH!** Switching gives you a 2/3 chance of winning.

**Why?**
- Your initial choice has a 1/3 chance of being correct
- The car must be behind either your door OR the remaining door
- Therefore, the remaining door has a 2/3 chance of being correct

Let's **prove it with simulation**!

```python
def play_monty_hall(switch=True):
    """
    Simulate one game of Monty Hall.

    Parameters:
    switch (bool): If True, contestant switches doors. If False, stays with original choice.

    Returns:
    bool: True if contestant wins the car, False otherwise
    """
    # Randomly place the car behind door 0, 1, or 2
    car_door = np.random.choice([0, 1, 2])

    # Contestant makes initial choice
    contestant_choice = np.random.choice([0, 1, 2])

    # Host opens a door (not car, not contestant's choice)
    remaining_doors = [d for d in [0, 1, 2] if d != car_door and d != contestant_choice]
    if remaining_doors:
        host_opens = np.random.choice(remaining_doors)
    else:
        # Contestant chose the car, host can open either goat door
        host_opens = np.random.choice([d for d in [0, 1, 2] if d != contestant_choice])

    # Switching strategy
    if switch:
        # Switch to the other unopened door
        final_choice = [d for d in [0, 1, 2] if d != contestant_choice and d != host_opens][0]
    else:
        # Stay with original choice
        final_choice = contestant_choice

    # Did we win?
    return final_choice == car_door

# Test it once
print(f"Switching - Won? {play_monty_hall(switch=True)}")
print(f"Staying - Won? {play_monty_hall(switch=False)}")
```

```python
# Simulate 10,000 games with each strategy
num_games = 10000

# Strategy 1: Always switch
switch_wins = []
for i in range(num_games):
    won = play_monty_hall(switch=True)
    switch_wins.append(won)

# Strategy 2: Never switch
stay_wins = []
for i in range(num_games):
    won = play_monty_hall(switch=False)
    stay_wins.append(won)

switch_wins = np.array(switch_wins)
stay_wins = np.array(stay_wins)

# Calculate win rates
switch_win_rate = np.mean(switch_wins)
stay_win_rate = np.mean(stay_wins)

print(f"\nResults from {num_games} games:")
print(f"Win rate when SWITCHING: {switch_win_rate:.1%}")
print(f"Win rate when STAYING:   {stay_win_rate:.1%}")
print(f"\nSwitching is {switch_win_rate/stay_win_rate:.2f}x better!")
```

```python
# Visualize the results
strategies = ['Always Switch', 'Always Stay']
win_rates = [switch_win_rate, stay_win_rate]

plt.figure(figsize=(10, 6))
bars = plt.bar(strategies, win_rates, color=['green', 'red'], alpha=0.7, edgecolor='black')
plt.ylabel('Win Rate')
plt.title('Monty Hall Problem: Should You Switch?')
plt.ylim(0, 1)
plt.axhline(0.67, color='green', linestyle='--', alpha=0.5, label='Theoretical (2/3)')
plt.axhline(0.33, color='red', linestyle='--', alpha=0.5, label='Theoretical (1/3)')
plt.legend()

# Add percentage labels on bars
for bar, rate in zip(bars, win_rates):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{rate:.1%}',
             ha='center', va='bottom', fontsize=14, fontweight='bold')

plt.show()
```

# Part 9: Basic Probability Rules

### What is Probability?

**Probability** is a number between 0 and 1 (or 0% to 100%) that measures how likely an event is to occur.

- **Probability = 0**: Impossible (never happens)
- **Probability = 1**: Certain (always happens)
- **Probability = 0.5**: Equally likely to happen or not

### Notation

We write $P(\text{event})$ to denote "the probability that event happens."

For example:
- $P(\text{heads on coin flip}) = 0.5$
- $P(\text{rolling a 6}) = \frac{1}{6} \approx 0.167$

### Rule 1: When an Event Doesn't Happen

$$P(\text{event doesn't happen}) = 1 - P(\text{event happens})$$

**Example**: If the probability of rain is 40%, then the probability of no rain is 60%.

```python
# Calculate complement
p_rain = 0.40
p_no_rain = 1 - p_rain
print(f"P(rain) = {p_rain}")
print(f"P(no rain) = {p_no_rain}")
```

### Rule 2: When All Outcomes are Equally Likely

$$P(\text{event}) = \frac{\text{number of favorable outcomes}}{\text{total number of outcomes}}$$

**Example**: Probability of rolling an even number on a die:

```python
# Define the sample space
die_outcomes = np.array([1, 2, 3, 4, 5, 6])

# Define favorable outcomes (even numbers)
favorable = np.array([2, 4, 6])

# Calculate probability
p_even = len(favorable) / len(die_outcomes)
print(f"P(even number) = {len(favorable)}/{len(die_outcomes)} = {p_even}")
```

### Rule 3: The Addition Rule (OR)

For **mutually exclusive** events (events that cannot happen at the same time):

$$P(A \cup B) = P(A) + P(B)$$

**Example**: Rolling a 1 OR a 6:

```python
p_roll_1 = 1/6
p_roll_6 = 1/6
p_roll_1_or_6 = p_roll_1 + p_roll_6

print(f"P(rolling 1) = {p_roll_1:.3f}")
print(f"P(rolling 6) = {p_roll_6:.3f}")
print(f"P(rolling 1 OR 6) = {p_roll_1_or_6:.3f}")
```

### Rule 4: The Multiplication Rule (AND)

For **independent** events (events that don't affect each other):

$$P(A \text{ AND } B) = P(A) \times P(B)$$

**Example**: Flipping heads AND rolling a 6:

```python
p_heads = 0.5
p_six = 1/6
p_heads_and_six = p_heads * p_six

print(f"P(heads) = {p_heads}")
print(f"P(rolling 6) = {p_six:.3f}")
print(f"P(heads AND rolling 6) = {p_heads_and_six:.3f}")
```

# Part 10: Frequentist vs. Bayesian Thinking

### Two Philosophies of Probability

#### Frequentist Approach
- Probability = **long-run frequency**
- "If I repeat this experiment infinitely many times, what proportion will succeed?"
- Example: "The probability of heads is 0.5 because in infinite flips, half will be heads."

#### Bayesian Approach  
- Probability = **degree of belief**
- Updates beliefs as new evidence arrives
- Example: "Based on my knowledge, I believe there's a 0.5 probability of heads."

### Demonstrating Frequency Interpretation

Let's show that in the long run, the proportion of heads approaches 0.5:

```python
# Simulate many coin flips and track cumulative proportion
num_flips = 1000
coin = np.array(['Heads', 'Tails'])
flips = np.random.choice(coin, size=num_flips)

# Calculate cumulative proportion of heads
cumulative_heads = np.cumsum(flips == 'Heads')
cumulative_proportion = cumulative_heads / np.arange(1, num_flips + 1)

# Plot
plt.figure(figsize=(12, 6))
plt.plot(cumulative_proportion, alpha=0.7)
plt.axhline(0.5, color='red', linestyle='--', linewidth=2, label='True probability (0.5)')
plt.xlabel('Number of Flips')
plt.ylabel('Proportion of Heads')
plt.title('Law of Large Numbers: Proportion of Heads Converges to 0.5')
plt.legend()
plt.ylim(0.3, 0.7)
plt.grid(alpha=0.3)
plt.show()

print(f"After {num_flips} flips, proportion of heads: {cumulative_proportion[-1]:.4f}")
```

#### O. J. Simpson: The Other Side of Probability

* Alan Dershowitz used the prosecutor's fallacy in the O.J. Simpson trial.
* The prosecution focused on Simpson's history of violence against Nicole.
* Dershowitz argued that most abusive spouses don't kill their partners, which is true.
* However, he ignored the statistic that most murdered battered women are killed by their abusers.
* The jury found Simpson not guilty.
* Dershowitz believes the justice system doesn't require telling the whole truth.


This scenario is actually a classic combination of the **Base Rate Fallacy** and the **Prosecutor’s Fallacy**, though it specifically highlights how "relevance" can be manipulated by shifting the base rate.

In the O.J. Simpson trial, Alan Dershowitz’s argument relied on using a broad, irrelevant base rate to mask a much more specific and relevant probability.

### Why it is a Base Rate Fallacy

The **Base Rate Fallacy** occurs when we ignore specific information and focus instead on a general statistic that doesn't apply to the current context.

* **The General Base Rate (Dershowitz's Argument):** Out of all men who abuse their partners, only about 1 in 2,500 eventually murder them. This makes the probability of murder seem incredibly low.
* **The Specific Base Rate (The Missing Context):** We are not looking at all battered women; we are looking at a **battered woman who has already been murdered**.
* **The Correct Statistic:** In cases where a woman is battered and then murdered, the batterer is the killer more than **80%** of the time.

By focusing on the "1 in 2,500" figure, the defense invited the jury to ignore the most important piece of evidence: that Nicole Brown Simpson was already dead.

### Comparison of the Two Perspectives

| Perspective | Question Being Asked | Base Rate Used | Result |
| --- | --- | --- | --- |
| **Defense** | What is the chance an abuser kills his wife? | All abusive relationships. | Extremely Low (~0.04%) |
| **Prosecution** | Given she was murdered, who is the likely killer? | Battered women who are murdered. | Extremely High (>80%) |


### Comparing Conditional Probabilities: The Dershowitz Argument

We can use **Bayes' Theorem** to show the difference between the probability the defense presented and the probability that was actually relevant to the case.

**Remember Conditional Probability**

$P(A|B) = \frac{P(A \cap B)}{P(B)}$


**Bayes Theorem**

$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$

# Mutual Exclusivity

### 1. Reminder: intersection is the key object

Both conditional probability and Bayes’ Theorem are built on the **intersection** (A \cap B):

$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$

$
P(A \cap B) = P(B \mid A) P(A)
$

Bayes’ Theorem exists *only because* this intersection can be non-zero.

### 2. What mutual exclusivity means

Two events (A) and (B) are **mutually exclusive** if they cannot occur together:

$
A \cap B = \varnothing
\quad \Longrightarrow \quad
P(A \cap B) = 0
$

### 3. Effect on conditional probability

Substitute into the definition:

$
P(A \mid B)
= \frac{P(A \cap B)}{P(B)}
= \frac{0}{P(B)}
= 0
\quad \text{(provided } P(B) > 0\text{)}
$

Interpretation:

If (B) occurred, then (A) **cannot** have occurred.

### 4. Effect on Bayes’ Theorem

Bayes’ Theorem:

$
P(A \mid B) = \frac{P(B \mid A) P(A)}{P(B)}
$

If (A) and (B) are mutually exclusive:

$
P(B \mid A) = 0
$

so

$
P(A \mid B)
= \frac{0 \cdot P(A)}{P(B)}
= 0
$

Bayes’ Theorem still holds algebraically — but it becomes **trivial**.

### 5. Why Bayes becomes uninformative under mutual exclusivity

Bayes’ Theorem is useful when:

$
P(A \cap B) > 0
$

Mutual exclusivity forces:

$
P(A \cap B) = 0
$

so no updating is possible — the posterior collapses to zero.

This is why Bayes is about **overlapping hypotheses**, not disjoint ones.

### 6. Contrast with independence (important distinction)

| Relationship       | Intersection             |
| ------------------ | ------------------------ |
| Mutually exclusive | (P(A \cap B) = 0)        |
| Independent        | (P(A \cap B) = P(A)P(B)) |

* **Mutual exclusivity** kills Bayes updates
* **Independence** simplifies Bayes updates

### 7. Summary chain (clean and complete)

$$
\begin{aligned}
P(A \mid B) &= \frac{P(A \cap B)}{P(B)} \\
P(A \cap B) &= P(B \mid A)\,P(A) \\
A \cap B = \varnothing &\Rightarrow P(A \cap B) = 0 \Rightarrow P(A \mid B) = 0
\end{aligned}
$$

* “The probability of A given B equals the probability that A and B both occur, divided by the probability of B.”
* “The probability that A and B both occur equals the probability of B given A times the probability of A.”
* “If A and B are mutually exclusive, meaning they cannot occur together, then the probability that both occur is zero.”
* “Therefore, the probability of A given B is zero.”

### 8. One-line intuition

Bayes’ Theorem requires **shared probability mass**; mutual exclusivity removes it entirely.

While the defense (Dershowitz) used a broad, irrelevant base rate to make the defendant seem innocent, the Prosecutor’s Fallacy occurs when a statistic is used to misrepresent the probability of innocence.
